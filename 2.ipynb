```python
import pandas as pd

df = pd.read_csv("answer.csv")

df = (df.groupby("query_id", as_index=False)["answer"]
        .apply(lambda x: " ".join(x.astype(str))))

df = df[["query_id", "answer"]]
df.to_csv("answer.csv", index=False)

# Проверка
print(df.head())
print("Уникальных query_id:", df["query_id"].nunique())
print("Всего строк:", len(df))
```
